```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# --- 1. SETUP RAW DATA ---
experience_years = [float(x) for x in "1.0,2.5,4.0,5.5,7.0,10.0".split(",")]
salaries_usd = [float(x) for x in "45000,58000,72000,85000,98000,125000".split(",")]

X = pd.DataFrame(experience_years, columns=['Years_Experience'])
y = pd.Series(salaries_usd)

# --- 2. TRAIN REGRESSION ENGINE ---
salary_model = LinearRegression()
salary_model.fit(X, y)

# --- 3. GENERATE PREDICTIONS FOR THE LINE ---
# Generate predictions for every row to chart the trend coordinates
y_pred = salary_model.predict(X)


# ==========================================
# === MATPLOTLIB VISUALIZATION PASSTHROUGH ===
# ==========================================
plt.figure(figsize=(9, 5))

# Plot Step A: Render the actual data records as a Scatter Plot
plt.scatter(X, y, color='crimson', marker='o', s=80, label='Actual Employee Records')

# Plot Step B: Render the AI's learned regression line straight through the points
plt.plot(X, y_pred, color='royalblue', linewidth=2.5, label='AI Regression Line (ŷ = wx + b)')

# --- 4. FORMAT HEADERS & LABELS ---
plt.title('HR Payroll Analysis: Experience vs Salary USD', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Years of Professional Experience', fontsize=11)
plt.ylabel('Salary Package ($)', fontsize=11)

# Format the gridlines and activate the legend box cleanly
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=10, loc='upper left')

# Render the canvas viewport pop-up window
plt.tight_layout()
plt.show()
```
