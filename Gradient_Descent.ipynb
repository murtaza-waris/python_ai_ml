```python
import numpy as np

# Sample data matching your kernel state
X = np.array([[100.0, 300.0], [50.0, 200.0], [0.0, 100.0]])
y = np.array([500.0, 350.0, 200.0])

# Standardize features for gradient descent stability
X_mean = np.mean(X, axis=0)
X_std = np.std(X, axis=0)
X_scaled = (X - X_mean) / X_std

# Hyperparameters
learning_rate = 0.01
epochs = 1000

# Initialization
n_samples, n_features = X_scaled.shape
weights = np.zeros(n_features)
bias = 0.0

# Gradient Descent loop
for epoch in range(epochs):
    # Model predictions
    y_pred = np.dot(X_scaled, weights) + bias

    # Calculate gradients
    dw = (1 / n_samples) * np.dot(X_scaled.T, (y_pred - y))
    db = (1 / n_samples) * np.sum(y_pred - y)

    # Update weights and bias
    weights -= learning_rate * dw
    bias -= learning_rate * db

    # Print cost occasionally
    if epoch % 100 == 0:
        cost = (1 / (2 * n_samples)) * np.sum((y_pred - y) ** 2)
        print(f"Epoch {epoch}: Cost = {cost:.4f}")

print(f"\nFinal weights: {weights}")
print(f"Final bias: {bias}")
```
