```python
from abc import ABC, abstractmethod

class BaseLayer(ABC):
    def __init__ (self, layer_name):
        self.name = layer_name

    @abstractmethod
    def forward (self, input_vector):
        pass

class DenseLayer (BaseLayer):
    def forward (self, input_vector):
        input_vector = input_vector*2
        return f"Calculation result: {input_vector}"

my_layer = DenseLayer("Hidden_Layer_01")
print(f"Layer Registered: {my_layer.name}")

output_signal = my_layer.forward(10.0)
print(f"Forward Pass Output Signal: {output_signal}")
```

```python
from abc import ABC, abstractmethod

# 1. Define the Abstract Parent Class Framework
class BaseCleaner(ABC):

    # A regular method that all child classes inherit automatically
    def count_raw_chars(self, text):
        return len(text)

    # Add the abstract decorator and method template here
    @abstractmethod
    def clean_text(self, text):
        pass


# 2. Define the Concrete Child Class (Make sure it inherits from BaseCleaner!)
class SpaceCleaner(BaseCleaner):

    # Fulfill the abstract contract by overriding the clean_text method
    def clean_text(self, text):
        # Clean the text using .strip() and return it
        clean_text = text.strip()
        return clean_text

my_cleaner = SpaceCleaner()

raw_string = "   Wah Cantt, Pakistan   "

# Test A: Run the inherited standard parent method
raw_count = my_cleaner.count_raw_chars(raw_string)
print(f"Total Raw Characters: {raw_count}") # Includes all spaces!

# Test B: Run your overridden abstract method contract
clean_string = my_cleaner.clean_text(raw_string)
print(f"Cleaned Text Output: '{clean_string}'")
print(f"Cleaned Characters: {len(clean_string)}")
```
