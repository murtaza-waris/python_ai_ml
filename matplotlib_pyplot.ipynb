```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Marketing analytics summary
report_data = {
    'Device_Platform': ['Mobile', 'Desktop', 'Tablet'],
    'Total_Revenue_USD': [15400.50, 28900.00, 4200.75],
    'Average_Conversion_Rate': [4.2, 5.8, 2.1]
}

df_report = pd.DataFrame(report_data)
print("--- Baseline Summary Report ---")
print(df_report)

#  Task 2: Build a Beautiful Bar Chart
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))
sns.barplot(data=df_report, x='Device_Platform',
            y='Total_Revenue_USD', hue='Device_Platform',
            palette='magma', legend=False)

#Task 3: Annotate and Display Your Matrix Graph
plt.title('Total Revenue Generation by Device Platform', fontsize=14, fontweight='bold')
plt.xlabel('Device Platform', fontsize=12, labelpad=10)
plt.ylabel('Total Revenue ($)', fontsize=12, labelpad=10)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Simulated performance metrics for an AI server cluster
metrics_data = {
    'CPU_Utilization_Pct': [45.2, 88.1, 95.0, 31.5, 62.0, 15.1],
    'Memory_Usage_GB': [16.1, 31.8, 32.0, 8.5, 24.1, 4.0],
    'Request_Latency_ms': [12.5, 45.1, 55.2, 8.1, 22.0, 4.2],
    'Server_Temperature_C': [42.1, 65.5, 71.0, 38.0, 51.5, 35.2]
}

df_cluster = pd.DataFrame(metrics_data)
print("--- Baseline Server Cluster Metrics ---")
print(df_cluster)

#Task 1: Generate the Core Numerical Matrix
corr_matrix = df_cluster.corr()
print("\n--- Correlation Matrix ---")
print(corr_matrix)

# Task 2: Configure a Correlation Heatmap Canvas
plt.figure(figsize=(6,6))
sns.heatmap(data=corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)

#Task 3: Header Annotation and Display
plt.title('AI Server Cluster Correlation Matrix', fontsize=14, fontweight='bold', pad=15)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- 1. SETUP COMPLETE DATA (Bypassing markdown stripping glitches) ---
hours_list = [int(x) for x in "1,2,3,4,5,6,7,8".split(",")]
model_a_list = [int(x) for x in "120,150,450,480,210,180,510,550".split(",")]

time_series_data = {
    'Shift_Hour': hours_list,
    'Model_A_Requests': model_a_list,
    'Model_B_Requests': [310, 290, 280, 300, 320, 310, 295, 305]
}

df_timeline = pd.DataFrame(time_series_data)
print("--- Baseline Time-Series Shift Logs ---")
print(df_timeline)

# Task 1: Establish Your Canvas Theme
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10,5))

#  Task 2: Plot Dual Trend Lines Simultaneously
sns.lineplot(data=df_timeline, x='Shift_Hour', y='Model_A_Requests', label="Model A (NLP Engine)",  color='royalblue', marker='o', linewidth=2.5)
sns.lineplot(data=df_timeline, x='Shift_Hour', y='Model_B_Requests', label="Model B (Vision Engine)",  color='darkorange', marker='o', linewidth=2.5)

#  Task 3: Visual Polish and Legend Activation
plt.title('AI Model Request Volume Ingestion Over 8-Hour Shift', fontsize=14, fontweight='bold', pad=15)
plt.xlabel ('Hour of Shift', fontsize=12)
plt.ylabel('Number of Requests', fontsize=12)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. SETUP COMPREHENSIVE TRAINING METRICS ---
epochs = [int(x) for x in "1,2,3,4,5,6".split(",")]
training_loss = [float(x) for x in "0.95,0.62,0.41,0.28,0.19,0.11".split(",")]
model_accuracy = [float(x) for x in "55.2,71.0,82.5,89.1,93.4,97.8".split(",")]


# --- 2. INITIALIZE MULTI-AXES SUBPLOT GRID (2 Rows, 1 Column) ---
# Setting nrows=2 and ncols=1 stacks the plots vertically
# figsize updated to be taller (7 inches wide, 10 inches high)
fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(7, 10))


# --- 3. PLOT TOP CANVAS (Row Index 0): LOSS OPTIMIZATION CURVE ---
axes[0].plot(epochs, training_loss, color='crimson', marker='o', linewidth=2.5)

axes[0].set_title('Model Loss Optimization Curve', fontsize=12, fontweight='bold', pad=10)
axes[0].set_xlabel('Training Epoch', fontsize=10)
axes[0].set_ylabel('Loss Value', fontsize=10)
axes[0].grid(True, linestyle=':', alpha=0.6)


# --- 4. PLOT BOTTOM CANVAS (Row Index 1): MODEL TRAINING ACCURACY ---
axes[1].plot(epochs, model_accuracy, color='forestgreen', marker='s', linewidth=2.5)

axes[1].set_title('Model Validation Accuracy Trend', fontsize=12, fontweight='bold', pad=10)
axes[1].set_xlabel('Training Epoch', fontsize=10)
axes[1].set_ylabel('Accuracy Percentage (%)', fontsize=10)
axes[1].grid(True, linestyle=':', alpha=0.6)


# --- 5. VISUAL MARGIN POLISH AND DISPLAY ---
# This is even MORE important now to keep the top x-label from crashing into the bottom title!
plt.tight_layout()
plt.show()
```
