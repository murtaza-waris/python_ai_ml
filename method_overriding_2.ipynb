```python
class SimplePrinter:
    def __init__ (self, msg):
        self.msg = msg

    def display(self):
        return self.msg

class ShoutingPrinter (SimplePrinter):

    def display(self):
        self.msg = self.msg.upper()
        return self.msg

class BracketPrinter (SimplePrinter):

    def display(self):
        return f"[{self.msg}]"

shouter = ShoutingPrinter("hello World")
bracketer = BracketPrinter("hello World")

print(shouter.display())
print(bracketer.display())
```

```python
class Vehicle:
    def __init__(self, vehicle_id):
        self.vehicle_id = vehicle_id

class ElectricTruck (Vehicle):
    def __init__(self, vehicle_id, battery_range):
        super().__init__(vehicle_id)
        self.range = battery_range

    def show_truck_specs(self):
        print(f"Truck ID: {self.vehicle_id} | Maximum Range: {self.range} km")


veh = Vehicle ("any")
truc = ElectricTruck ("bony", 100)

truc.show_truck_specs()
```

```python
class SecurityGate:

    def open_gate(self):
        print("🔓 Lock mechanism disengaged. Gate opened.")

class BioMetricGate (SecurityGate):

    def open_gate(self):
        super().open_gate()
        print("📸 Camera active: Biometric face log saved to security database.")

gate = BioMetricGate()
gate.open_gate()

```
