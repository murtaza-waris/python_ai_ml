```python
x = int(input("Provide first number: "))
y = int (input("Provide second number: "))

try:
    print ("Database connection started")
    result = x/y
    p = int(input("Provide another number: "))
    print (result)
    print (p)

except ZeroDivisionError as e:
    print("you cannot divide by 0", e)

except ValueError as e:
    print("please enter a number", e)
except Exception as e:
    print("something wrong with code", e)


finally:
    print ("Database connection closed")


print("Next line of code")
```
