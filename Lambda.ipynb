```python
#Lambda function

#Write a single-line lambda function that takes a number and divides it by 100.0.
#Then use it with Python's built-in map() tool to scale down a whole list of sensor values.

List = [232,456,670,894,638]

newList = list(map(lambda n: n/100.0, List))
print(newList)
```

```python
sorted_list = sorted(newList, key=lambda x: x)
print("Sorted list:", sorted_list)
```

```python
newList2 = list(filter(lambda n: n%2 == 0, sorted_list))
print(newList2)
```

```python
from functools import reduce
total = reduce(lambda x,y:x+y, sorted_list)
print (total)
```
