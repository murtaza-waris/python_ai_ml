```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("/content/drive/MyDrive/Pandas/Experience-Salary.csv")

df.columns = ["exp_in_months","salary_in_thousands"]
df = df[df['salary_in_thousands'] >= 1].reset_index(drop=True)
df = df.drop_duplicates()
df
```

```python
plt.scatter(df.exp_in_months,df.salary_in_thousands)
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()
```

```python
salary_model = linear_model.LinearRegression()
salary_model.fit(df[['exp_in_months']],df.salary_in_thousands)
```

```python
salary_model.predict([[20.258754]])
```

```python
salary_model.score(df[['exp_in_months']],df.salary_in_thousands)
```

```python
salary_model.coef_
```

```python
salary_model.intercept_
```
