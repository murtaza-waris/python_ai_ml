```python
import numpy as np

# Image Patches: 2 samples, 3 pixels each. Shape: (2, 3)
image_patches = np.array([
    [1.0, 0.5, 0.2],
    [0.0, 0.8, 1.0]
], dtype=np.float32)

# Detector Matrix: 3 inputs mapped onto 2 distinct pattern filters. Shape: (3, 2)
detectors = np.array([
    [0.5, -0.5],
    [1.0,  0.0],
    [0.1,  0.9]
], dtype=np.float32)

print("Image Patches shape: ", image_patches.shape)
print("Detector Matrix shape: ", detectors.shape)

extracted_features = image_patches @ detectors
print("Extracted Features shape: ", extracted_features.shape)
print(extracted_features)

first_column_index = extracted_features[:, 0]
print(first_column_index)
```
