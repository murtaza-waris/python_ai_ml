```python
class LanguagueModelWorker:
    def __init__ (self, model_name):
        self.name = model_name

    def get_metrics(self):
        return f"[LLM: {self.name}] RAM usage: 14.2 GB | Requests/sec: 125"

class VisionClusterWorker:
    def __init__ (self, cluster_id):
        self.id = cluster_id

    def get_metrics(self):
        return f"[Vision: {self.id}] GPU Temp: 68°C | VRAM used: 92%"

def log_system_diagnostics(worker_object):
    result = worker_object.get_metrics()
    print(result)

llm_node = LanguagueModelWorker("BERT")
vision_node = VisionClusterWorker("A100")

log_system_diagnostics(llm_node)
print()
log_system_diagnostics(vision_node)
```
